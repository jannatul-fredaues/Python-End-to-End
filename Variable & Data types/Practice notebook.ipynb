```python
#Remainder is negative when denominator is negative
# n:   + - + -
# d:   + - - +
# ans: + - - +

#case: 1
a,b = 5, 2
c = a%b
print(c)

#case: 2
i,j = -5,-2
k = i%j
print(k)

#case: 3
m,n= 5,-2
o = m%n
print(o)

#case: 4
x,y = -5,2
z = x%y
print(z)
```
